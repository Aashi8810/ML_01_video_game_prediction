```python
import joblib
import pandas as pd
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt

# Load your encoded test set and trained model
X_test = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/X_test_encoded.csv')
y_test = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/y_test_encoded.csv')
best_svr = joblib.load('svr_model.pkl')

# Compute permutation importance
result = permutation_importance(best_svr, X_test, y_test, n_repeats=10, random_state=42, scoring='r2')

# Create dataframe
importance_df = pd.DataFrame({
    'Feature': X_test.columns,
    'Importance': result.importances_mean
}).sort_values(by='Importance', ascending=False)

# Display and plot
print("\n🎯 Top 15 Most Influential Features:")
print(importance_df.head(15))

plt.figure(figsize=(8, 6))
plt.barh(importance_df.head(15)['Feature'][::-1], importance_df.head(15)['Importance'][::-1])
plt.xlabel("Mean Importance (ΔR² when permuted)")
plt.title("Top 15 Feature Importances (SVR via Permutation)")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load encoded data (so we have all features)
X_train = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/X_train_encoded.csv')
y_train = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/y_train_encoded.csv')

# Combine features with target for correlation analysis
df_corr = pd.concat([X_train, y_train.rename(columns={'Global_Sales_Log': 'Target'})], axis=1)

# Compute correlation matrix
corr_matrix = df_corr.corr()

# Focus on top correlations with the target variable
top_corr = corr_matrix['Target'].sort_values(ascending=False).head(15)

# Display top correlations
print("\n🔥 Top 15 features most correlated with Global Sales:")
print(top_corr)

# Create heatmap for the top 15 correlated features
plt.figure(figsize=(10, 8))
sns.heatmap(df_corr[top_corr.index].corr(), annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title("Heatmap of Top 15 Features Most Correlated with Global Sales")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load full encoded data
X_train = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/X_train_encoded.csv')
y_train = pd.read_csv('/workspaces/Project_01/video_game_analysis/notebooks/FC110542_Weerakoon/Processed data/y_train_encoded.csv')

# Merge target
df_corr = pd.concat([X_train, y_train.rename(columns={'Global_Sales_Log': 'Target'})], axis=1)

# --- 1️⃣ GENRE features ---
genre_cols = [col for col in df_corr.columns if col.startswith('Genre_')]
plt.figure(figsize=(10, 6))
sns.heatmap(df_corr[genre_cols + ['Target']].corr()[['Target']].sort_values(by='Target', ascending=False), annot=True, cmap='coolwarm')
plt.title("🎮 Correlation of Genres with Global Sales")
plt.tight_layout()
plt.show()

# --- 2️⃣ PLATFORM features ---
platform_cols = [col for col in df_corr.columns if col.startswith('Platform_')]
plt.figure(figsize=(10, 6))
sns.heatmap(df_corr[platform_cols + ['Target']].corr()[['Target']].sort_values(by='Target', ascending=False), annot=True, cmap='coolwarm')
plt.title("🕹️ Correlation of Platforms with Global Sales")
plt.tight_layout()
plt.show()

# --- 3️⃣ PUBLISHER features ---
publisher_cols = [col for col in df_corr.columns if col.startswith('Publisher_')]
plt.figure(figsize=(10, 8))
sns.heatmap(df_corr[publisher_cols + ['Target']].corr()[['Target']].sort_values(by='Target', ascending=False).head(15), annot=True, cmap='coolwarm')
plt.title("🏢 Top 15 Publishers Most Correlated with Global Sales")
plt.tight_layout()
plt.show()
```
