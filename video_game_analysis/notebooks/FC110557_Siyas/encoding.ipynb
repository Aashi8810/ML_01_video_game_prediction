```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Load splits
X_train = pd.read_csv("dataSplits/X_train.csv")
X_val = pd.read_csv("dataSplits/X_val.csv")
X_test = pd.read_csv("dataSplits/X_test.csv")

# Detect object (categorical) columns
cat_cols = X_train.select_dtypes(include='object').columns.tolist()

# Apply LabelEncoder
for col in cat_cols:
    le = LabelEncoder()
    
    # Fit on training
    le.fit(X_train[col])
    
    # Add 'Unknown' to known classes
    if 'Unknown' not in le.classes_:
        le.classes_ = np.append(le.classes_, 'Unknown')

    # Transform train
    X_train[col] = le.transform(X_train[col])
    
    # Map unseen to 'Unknown', then transform - To safely encode unseen categories in validation/test sets using the training encoder only
    X_val[col] = le.transform(X_val[col].map(lambda s: s if s in le.classes_ else 'Unknown').fillna('Unknown'))
    X_test[col] = le.transform(X_test[col].map(lambda s: s if s in le.classes_ else 'Unknown').fillna('Unknown'))


# Save the label-encoded versions
X_train.to_csv("dataSplits/X_train_labeled.csv", index=False)
X_val.to_csv("dataSplits/X_val_labeled.csv", index=False)
X_test.to_csv("dataSplits/X_test_labeled.csv", index=False)
```
